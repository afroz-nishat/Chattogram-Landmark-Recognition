```python
import requests
import time
import json
import pandas as pd
from pathlib import Path

API_URL = "https://commons.wikimedia.org/w/api.php"

HEADERS = {
    "User-Agent": (
        "ChattogramDivisionLandmarkDataset/1.0 "
        "(research dataset collection; educational use)"
    )
}

session = requests.Session()
session.headers.update(HEADERS)

CACHE_DIR = BASE / "07_cache" / "wikimedia_scan"
CACHE_DIR.mkdir(parents=True, exist_ok=True)

print("Wikimedia scanner ready.")
```

```python
def wikimedia_request(params, max_retries=6):
    """
    Make a Wikimedia API request with exponential backoff.
    Handles rate limits and temporary server errors.
    """

    delay = 2

    for attempt in range(max_retries):

        try:
            response = session.get(
                API_URL,
                params=params,
                timeout=60
            )

            if response.status_code == 200:
                return response.json()

            if response.status_code in [429, 500, 502, 503, 504]:
                print(
                    f"Temporary HTTP {response.status_code}. "
                    f"Waiting {delay}s..."
                )
                time.sleep(delay)
                delay *= 2
                continue

            response.raise_for_status()

        except requests.RequestException as e:

            print(
                f"Request error: {e}. "
                f"Waiting {delay}s..."
            )

            time.sleep(delay)
            delay *= 2

    raise RuntimeError(
        "Wikimedia request failed after retries."
    )
```

```python
def search_commons(query, max_results=200):
    """
    Search Wikimedia Commons for image files.

    Returns:
        {
            "total_hits": ...,
            "files": [...]
        }
    """

    files = {}
    total_hits = None
    continue_params = {}

    while len(files) < max_results:

        params = {
            "action": "query",
            "generator": "search",
            "gsrsearch": query,
            "gsrnamespace": 6,
            "gsrlimit": 50,

            "prop": "imageinfo",
            "iiprop": "url|size|mime|extmetadata",
            "iiurlwidth": 1200,

            "format": "json",
            "formatversion": 2
        }

        params.update(continue_params)

        data = wikimedia_request(params)

        if total_hits is None:
            total_hits = (
                data.get("query", {})
                    .get("searchinfo", {})
                    .get("totalhits")
            )

        pages = (
            data.get("query", {})
                .get("pages", [])
        )

        if not pages:
            break

        for page in pages:

            page_id = page.get("pageid")

            if page_id is not None:
                files[page_id] = page

            if len(files) >= max_results:
                break

        if "continue" not in data:
            break

        continue_params = data["continue"]

        time.sleep(1.5)

    return {
        "total_hits": total_hits or 0,
        "files": list(files.values())
    }
```

```python
def scan_landmark(landmark, info):

    cache_file = CACHE_DIR / f"{landmark}.json"

    # Use cached result if available
    if cache_file.exists():

        with open(cache_file, "r", encoding="utf-8") as f:
            return json.load(f)

    all_files = {}

    query_results = []

    for query in info["queries"]:

        print(f"  Searching: {query}")

        try:

            result = search_commons(
                query,
                max_results=200
            )

            query_results.append({
                "query": query,
                "total_hits": result["total_hits"],
                "retrieved": len(result["files"])
            })

            for file in result["files"]:

                pageid = file.get("pageid")

                if pageid:
                    all_files[str(pageid)] = file

        except Exception as e:

            print("  ERROR:", e)

        # Important: slow down requests
        time.sleep(2)

    output = {
        "landmark": landmark,
        "district": info["district"],
        "queries": query_results,
        "unique_files_retrieved": len(all_files),
        "files": list(all_files.values())
    }

    with open(cache_file, "w", encoding="utf-8") as f:
        json.dump(
            output,
            f,
            indent=2,
            ensure_ascii=False
        )

    return output
```

```python
from PIL import Image
from io import BytesIO
import hashlib

MIN_WIDTH = 300
MIN_HEIGHT = 300


def validate_image_bytes(content):

    try:

        image = Image.open(
            BytesIO(content)
        )

        image.verify()

        image = Image.open(
            BytesIO(content)
        )

        width, height = image.size

        if width < MIN_WIDTH or height < MIN_HEIGHT:
            return False, "too_small"

        if image.format not in [
            "JPEG",
            "PNG",
            "WEBP"
        ]:
            return False, "unsupported_format"

        return True, "valid"

    except Exception:
        return False, "corrupt"
```

```python
def download_image(url):

    try:

        response = session.get(
            url,
            timeout=60
        )

        if response.status_code != 200:
            return None, f"http_{response.status_code}"

        valid, reason = validate_image_bytes(
            response.content
        )

        if not valid:
            return None, reason

        return response.content, "valid"

    except Exception as e:

        return None, str(e)
```

```python
def get_metadata(page):

    imageinfo = (
        page.get("imageinfo", [{}])[0]
    )

    metadata = imageinfo.get(
        "extmetadata",
        {}
    )

    def meta_value(key):

        value = metadata.get(key, {})

        if isinstance(value, dict):
            return value.get("value", "")

        return str(value)

    return {
        "pageid": page.get("pageid"),
        "title": page.get("title", ""),
        "file_url": imageinfo.get("url", ""),
        "thumb_url": imageinfo.get("thumburl", ""),
        "description_url": imageinfo.get("descriptionurl", ""),
        "width": imageinfo.get("width"),
        "height": imageinfo.get("height"),
        "mime": imageinfo.get("mime"),
        "license": meta_value("LicenseShortName"),
        "artist": meta_value("Artist"),
        "credit": meta_value("Credit"),
        "date": meta_value("DateTimeOriginal")
    }
```

```python
import re
import pandas as pd
from tqdm.auto import tqdm


def safe_filename(text):

    text = re.sub(
        r"[^A-Za-z0-9._-]+",
        "_",
        text
    )

    return text[:120]


download_records = []


# Maximum candidate images per class
MAX_CANDIDATES = 120


for landmark, info in LANDMARKS.items():

    cache_file = CACHE_DIR / f"{landmark}.json"

    if not cache_file.exists():
        print("Skipping:", landmark)
        continue

    with open(
        cache_file,
        "r",
        encoding="utf-8"
    ) as f:

        scan_data = json.load(f)

    pages = scan_data.get(
        "files",
        []
    )

    if len(pages) == 0:
        continue

    output_dir = (
        CANDIDATE_DIR / landmark
    )

    print(
        f"\n{landmark}: "
        f"{len(pages)} candidate files"
    )

    successful = 0

    seen_sha256 = set()

    for page in tqdm(
        pages[:MAX_CANDIDATES],
        desc=landmark
    ):

        metadata = get_metadata(page)

        url = (
            metadata["thumb_url"]
            or metadata["file_url"]
        )

        if not url:
            continue

        content, status = download_image(url)

        if content is None:
            download_records.append({
                "landmark": landmark,
                "status": status,
                "title": metadata["title"]
            })
            continue

        sha256 = hashlib.sha256(
            content
        ).hexdigest()

        # Exact duplicate
        if sha256 in seen_sha256:
            continue

        seen_sha256.add(sha256)

        filename = (
            f"{successful+1:04d}_"
            f"{safe_filename(metadata['title'])}.jpg"
        )

        filepath = (
            output_dir / filename
        )

        try:

            image = Image.open(
                BytesIO(content)
            ).convert("RGB")

            image.save(
                filepath,
                "JPEG",
                quality=95
            )

            download_records.append({
                "landmark": landmark,
                "district": info["district"],
                "filename": filename,
                "path": str(filepath),
                "sha256": sha256,
                "title": metadata["title"],
                "file_url": metadata["file_url"],
                "description_url": metadata["description_url"],
                "license": metadata["license"],
                "artist": metadata["artist"],
                "width": metadata["width"],
                "height": metadata["height"],
                "status": "downloaded"
            })

            successful += 1

        except Exception as e:

            print(
                "Save error:",
                e
            )

        time.sleep(0.3)

    print(
        "Successfully downloaded:",
        successful
    )
```

```python
import imagehash
from PIL import Image


def calculate_phash(path):

    try:

        image = Image.open(path).convert(
            "RGB"
        )

        return str(
            imagehash.phash(image)
        )

    except Exception:
        return None
```

```python
def hamming_distance(hash1, hash2):

    return imagehash.hex_to_hash(
        hash1
    ) - imagehash.hex_to_hash(
        hash2
    )
```

```python
PHASH_THRESHOLD = 6
```

```python
from collections import defaultdict


def remove_near_duplicates(df, threshold=6):

    keep_indices = []

    for landmark, group in df.groupby(
        "landmark"
    ):

        selected_hashes = []

        for idx, row in group.iterrows():

            current_hash = row["phash"]

            if not current_hash:
                continue

            duplicate = False

            for existing_hash in selected_hashes:

                if hamming_distance(
                    current_hash,
                    existing_hash
                ) <= threshold:

                    duplicate = True
                    break

            if not duplicate:

                keep_indices.append(idx)
                selected_hashes.append(
                    current_hash
                )

    return df.loc[
        keep_indices
    ].reset_index(drop=True)


clean_df = remove_near_duplicates(
    download_df,
    threshold=PHASH_THRESHOLD
)

print(
    "Before:",
    len(download_df)
)

print(
    "After pHash deduplication:",
    len(clean_df)
)
```

```python
before_counts = (
    download_df
    .groupby("landmark")
    .size()
    .rename("before")
)

after_counts = (
    clean_df
    .groupby("landmark")
    .size()
    .rename("after")
)

dedup_report = pd.concat(
    [before_counts, after_counts],
    axis=1
).fillna(0)

dedup_report["removed"] = (
    dedup_report["before"]
    -
    dedup_report["after"]
)

dedup_report = (
    dedup_report
    .reset_index()
    .sort_values(
        "removed",
        ascending=False
    )
)

display(dedup_report)
```
